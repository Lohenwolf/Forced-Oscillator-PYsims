```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

t_max = 40          # Increase to lengthen the simulation
dt = 0.01           
t = np.arange(0, t_max, dt)


skip = 25           # Plots every 25th point, decrease to increase smoothness

w0 = 1.0
wf_rational = 1.5 * w0 
wf_irrational = np.sqrt(2) * w0 

def motion(w_forced, time_array):
    x = np.cos(w0 * time_array) + np.cos(w_forced * time_array)
    v = -w0 * np.sin(w0 * time_array) - w_forced * np.sin(w_forced * time_array)
    return x, v

x_rat, v_rat = motion(wf_rational, t)
x_irr, v_irr = motion(wf_irrational, t)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), dpi=60)
plt.close()

def setup_axis(ax, x_data, v_data, title):
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("v")
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_xlim(np.min(x_data)-0.5, np.max(x_data)+0.5)
    ax.set_ylim(np.min(v_data)-0.5, np.max(v_data)+0.5)
    line, = ax.plot([], [], lw=0.8, alpha=0.8)
    point, = ax.plot([], [], 'o', color='black', markersize=4)
    return line, point

line1, point1 = setup_axis(ax1, x_rat, v_rat, r"Rational ratio ($\omega_f / \omega_0 = 1.5$)" + "\nClosed periodic trajectory")
line2, point2 = setup_axis(ax2, x_irr, v_irr, r"Irrational ratio ($\omega_f / \omega_0 = \sqrt{2}$)" + "\nDense, closed, quasi-periodic trajectory")

line1.set_color('blue')
line2.set_color('red')

def init():
    line1.set_data([], [])
    point1.set_data([], [])
    line2.set_data([], [])
    point2.set_data([], [])
    return line1, point1, line2, point2

def update(frame):
    line1.set_data(x_rat[:frame], v_rat[:frame])
    line2.set_data(x_irr[:frame], v_irr[:frame])
    
    point1.set_data([x_rat[frame]], [v_rat[frame]])
    point2.set_data([x_irr[frame]], [v_irr[frame]])
    return line1, point1, line2, point2

anim = FuncAnimation(fig, update, init_func=init, 
                     frames=range(0, len(t), skip), 
                     interval=80, blit=True)

HTML(anim.to_jshtml())
```
